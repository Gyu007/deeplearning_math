# 4.1 데이터의 종류

## 4.1.1 명목형 자료
- **범주형 자료(Categorical Data)** $\approx$ **명목형 자료(Nominal Data)**:
    - 서로 다른 값들 사이의 순서가 없는 데이터
    - ex) 갈색, 파란색, 녹색 사이에는 특별한 순서 관계가 없음

## 4.1.2 순서형 자료
- 순서형 자료(Ordinal Data; 또는 서열 자료):
    - 순위 또는 순서가 있는 데이터 (단, 갑들의 차이에 어떤 수학적 의미가 있지는 않음)
    - ex) 설문조사에서 "동의함", "강하게 동의함"은 순서는 존재하나 후자가 전자보다 3배 강하다는 의미는 아님

## 4.1.3 구간 자료
- **구간 자료(Interval Data)** 에서는 값들의 차이에 의미가 없음
    - ex) 온도가 40°F, 80°F인 물이 있을 때 둘의 온도차는 40°F이지만 80°F가 40°F보다 두 배로 뜨겁다고 할 수는 없음.
    - ex) 이를 섭씨로 바꾸면 40°F는 4.4°C,  80°F는 26.7°C로 단위를 바꾸면 6배가 된다.

=> 즉, 구간 자료에서는 차이의 비율이 큰 의미가 없음


## 4.1.4 비율 자료
- **비율 자료(Ratio Data)**:
    - 차이가 의미 있을 뿐만이 아니라 '진정한 영점(zero point)'이 존재하는 데이터
    - ex) 높이(키)는 0이라고 하는 영점이 존재함
    - ex) 4.1.3에서 언급한 온도 단위 외에 켈빈(K)은 비율 자료로 0K는 온도가 아예 없음을 의미함

## 4.1.5 명목형 자료와 심층학습
- ex) 빨간색, 녹색, 파란색과 같은 색상 자료는 명목형 자료이다.
- ex) 명목형 자료에는 순사 없으므로, 각각 빨간색=1, 녹색=2, 파란색=3으로 변환하는 것은 X
- ex) 이는 파란색이 빨간색의 3배이다라고 이해할 수 있음
- 명목형 자료를 사용 시에는 구간이 의미를 가지는 형태의 데이터로 변환할 필요가 있음
- 이때 흔히 사용하는 방식이 **원핫 부호화(one-hot encoding)** 이다.
- **원핫 부호화**:
    - 하나의 명목 변수(Nominal Variable)를 하나의 벡터로 변환
    - 이 벡터의 성분(요소)들은 명목 변수가 가질 수 있는 값들과 1대 1로 대응됨

| 값 |  | 벡터 |
| :--- | :--- | :--- |
| 빨간색 | → | 1 0 0 |
| 녹색   | → | 0 1 0 |
| 초록색 | → | 0 0 1 |
