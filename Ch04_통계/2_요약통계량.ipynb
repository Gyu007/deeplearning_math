# 4.2 요약 통계량

## 4.2.1 평균과 중앙값
**산술 평균(Arithmetic Mean)**:
- 모든 수를 더한 후 그 개수로 나누어 계산한 평균
- 데이터 집합을 $\{x_0,\, x_1\, x_2,...,\,x_{n-1}\}$이라고 할 때,

    $\bar{x}=\frac{1}{n}\sum_{i}^{n-1}x_i$

- 위 식은 비가중(unweighted) 평균으로 각 값에는 $1/n$의 가중치가 부여되며,</br>
그 가중치들을 모두 더하면 1.0이 됨

**가중 평균(Weighted Mean):
- 요소마다 개별적인 가중치를 적용하여 계산한 평균
- 집합의 요소가 모두 중요하지 않으며, 더 중요한 것과 덜 중요한 것이 있음
- 이러한 경우에는 요소들에 서로 다른 가중치를 부여해야 함

    $\bar{x}=\sum_{i}^{n-1}ω_ix_i$

- 여기서 $ω_i$는 $x_i$에 부여된 가중치로 $\sum_{i}ω_i=1$을 만족함
- 이 가중치는 데이터 집합의 일부가 아니며, 다른 어디에서 구해야 하는 값임

### 4.2.1.1 기하 평균
**기하 평균(Geometric Mean)**:
- 두 양수 a와 b의 곱의 제곱근
$$
    \bar{x}=\sum_{i}^{n-1}ω_ix_i
$$

- 일반화하면, 양수 n개의 기하 평균은 그 곱의 n제곱근이다.
$$
    x_g=\sqrt[n]{x_0 x_1 x_2 \cdot \cdot \cdot x_{n-1}}
$$

- 이미지 처리 분야에서는 이미지의 잡음을 줄이는 필터로서 기하 평균을 사용함
- 심층 학습에서 모형을 평가하는데 사용하는 계량(metric) 중 하나인</br>
**매튜 상관계수(Matthews Correlation Coefficient , MCC)** 에 등장함
- MCC는 다른 두 계량인 정보도(informedness)와 유표도(markedness)의 기하평균이다

### 4.2.1.2 조화 평균
- 두 수 a와 b의 **조화 평균(Harmonic Mean)은 그 역수들의 산술 평균의 역수

$$
    \bar{x}_h=\bigg(
        \frac{1}{2}
        \bigg(
            \frac{1}{a}+\frac{1}{b}
        \bigg)
    \bigg)^{-1}
$$

- 일반적으로는 아래와 같다.

$$
    \bar{x}_h=\bigg(
        \frac{1}{n}\sum_i^{n-1}\frac{1}{x_i}
    \bigg)^{-1}=\frac{n}{
        \frac{1}{x_0} + \frac{1}{x_1} + ⋅⋅⋅ + \frac{1}{x_{n-1}}
    }
$$

- 심층 학습에서 조화 평균은 F1 Score에 등장함
- F1 Score는 재현율(Recall; 민감도)와 정밀도(Precision)의 조화 평균이다

$$
    \text{F1}=\bigg(
        \frac{1}{2}
        \bigg(
            \frac{1}{\text{recall}} + \frac{1}{\text{precision}}
        \bigg)
    \bigg)^{-1}
$$

- 그러나 이것으로 심층학습 모형을 평가하는 것은 좋지 않음

$$
    \text{recall}=\frac{\text{TP}}{\text{TP + FN}}
    \\
    \text{precision}=\frac{\text{TP}}{\text{TP + FP}}
$$

- TP는 참 양성의 수, FN은 거짓 음성의 수, FP는 거짓 양성의 수
- 위 값들은 모형 평가 시 사용한 Test set에서 구함
- 이진 분류의 평가에 중요한 TN은 참 음성, 그러나 F1 Score는 TN을 무시함
- 모형을 제대로 평가하려면 양성과 음성 분류 모두 고려해야 함

### 4.2.1.3 중앙값
**중앙값(median)**
- 데이터 집합의 중앙에 있는 값
- 값들을 수치 크기 순으로 정렬 후, 절반은 중앙값 위에 있고 절반은 중앙값 아래에 있음

$$
    X=\{
        55,63,65,37,74,71,73,87,69,44
    \}
    \longrightarrow
    \{
        37,44,55,63,65,69,71,73,74,87
    \}
$$

- 그러나, 위와 같이 요소들이 홀수가 아닌 짝수일때, 가운데 두 값의 산술 평균을 중앙값으로 사용함

$$
    x_{median}=\frac{65 + 69}{2}=67
$$

- 평균은 값들의 크기 자체에 민감, 중앙값은 값들의 순서에 민감

## 4.2.2 변동의 측도

### 4.2.2.1 편차 대 분산

- 데이터 집합의 변동을 측정하는 한 방법은 데이터 집합의 **범위(range)**,</br>
즉, 가장 큰 값과 가장 작은 값의 차이를 구하는 것이다
- 그러나 범위는 데이터 집합의 오직 극단적인 두 값만 반영함
- 범위보다는, 평균과 각 값의 차이들의 평균이 데이터 집합의 변동을 좀 더 말해준다.

**평균 편차(Mean Deviation)**
$$
    MD = \frac{1}{n}\sum_i^{n-1}|x_i - \bar{x}|
$$
- 각 값이 평균적으로 전체 평균과 얼마나 떨어져 있는가

**편향 표본 분산(Biased Sample Variance)**
$$
    s_n^2 = \frac{1}{n}\sum_i^{n-1}(x_i - \bar{x})^2
$$
- 데이터 집합의 각 값과 평균의 차이의 제곱들의 평균

**비편향(불편) 표본 분산(Unbiased Sample Variance)**
$$
    s^2 = \frac{1}{n-1}\sum_i^{n-1}(x_i - \bar{x})^2
$$
- $n$ 대신 $n-1$을 사용한 것을 베셀 보정(Bassels Correction)이라고 함
- 이 보정은 잔차(Residual)들의 자유도와 관련 있으며, 잔차는 데이터 집합의 각 값에서 평균을 뺀 나머지이다


````python
# 값이 10개인 표분에서는 두 갑ㄳ은 소수점 이하 둘째 자리까지만 같다

import numpy as np

n = 10
a = np.random.random(n)

print(f"편향 분산: {(1/n)*((a-a.mean())**2).sum()}")
print(f"비편향 분산: {(1/(n-1))*((a-a.mean())**2).sum()}")
````

````python
# 표본 크기 10 -> 10,000
# 두 값은 소수점 이하 넷째 자리까지 같다

n = 10000
a = np.random.random(n)

print(f"편향 분산: {(1/n)*((a-a.mean())**2).sum()}")
print(f"비편향 분산: {(1/(n-1))*((a-a.mean())**2).sum()}")
````

**중앙값 절대 편차(MAD)**
- 표준편차는 평균에 기조차 통계량이고 극단적인 값들에 더욱 더 민감함
- 데이터 집합의 극단적인 값들에 덜 민감한 지표로 중앙값 절대 편차(Median Absolute Deviation, MAD)가 있다

$$
    \text{MAD} = \text{median}(|X_i - median(X)|)
$$

```
def MAD(x):
    return np.median(np.abs(x-np.median(x)))
```

**평균 표준오차(Standard Error of the Mean, SEM)=표준오차**
- 표본 평균들의 집합의 표준편차
$$
    SE = \frac{s}{\sqrt{n}}
